# C01 — Perforated absorbing film

## Mathematical and physical model

The structure is nonmagnetic and periodic in the transverse plane. In every constant-$z$ slice the electric field satisfies
$$
\nabla\times\nabla\times\mathbf E-k_0^2\epsilon_r(\mathbf r)\mathbf E=0,\qquad
\mathbf E_t(\mathbf r)=\sum_{m,n}\mathbf e_{mn}(z)e^{i(\mathbf k_{\parallel,0}+\mathbf G_{mn})\cdot (x,y)},\qquad
\mathbf G_{mn}=2\pi(m/P_x,n/P_y).
$$
Match tangential $\mathbf E$ and $\mathbf H$ across the layer interfaces. In each exterior medium $j$, the outgoing order is propagating when
$k_{z,j,mn}^2=k_0^2\epsilon_j-|\mathbf k_{\parallel,0}+\mathbf G_{mn}|^2>0$.
The comparison uses the complex amplitudes normalized so $|a_{mn}^{q\leftarrow p}|^2$ is the incident-power fraction in an outgoing polarization channel.

Passive absorption is determined by $A=1-\sum_{m,n}R_{mn}-\sum_{m,n}T_{mn}$. The film has refractive index $1.30+0.03i$ and the holes are filled with air.

## Physical scenario

A 65 nm absorbing film on glass is perforated by a square array of 90 nm radius holes at 400 nm pitch. The wavelengths span 650–780 nm.

## Possible applications

Perforated absorbing coatings and patterned optical attenuators.

## Benchmark objective

Compare reflected, transmitted, and absorbed power along with the zero-order complex amplitudes.

## Reproduce

Use the project environment and select **Run All**. The notebook loads the reported settings and reruns points with convergence evidence from all four solvers. The native [Lumerical model](model.fsp) and [reconstruction script](model.lsf) use the first excitation in [case.yaml](case.yaml).

[Published comparisons and figures](../../../BENCHMARK_REPORT.md#c01)

```python
from IPython.display import display
import matplotlib.pyplot as plt

from rcwa_cv.case_loader import catalog
from rcwa_cv.notebook import case_setup, show_results, show_settings, show_versions, solve_sweep
from rcwa_cv.reporting.plots import geometry_figure

case_path = catalog("C01")[0]
spec, benchmark = case_setup(case_path)
results = {}
show_settings(benchmark)

figure = geometry_figure(case_path)
display(figure)
plt.close(figure)
```

## meent · NumPy CPU

```python
from rcwa_cv.solvers.meent_cpu import probe

if benchmark["comparisons"]["meent"]:
    print(probe())
results["meent"] = solve_sweep("meent", spec, benchmark)
```

## torcwa · CUDA GPU, complex128

```python
from rcwa_cv.solvers.torcwa_cuda import probe

if benchmark["comparisons"]["torcwa"]:
    print(probe())
results["torcwa"] = solve_sweep("torcwa", spec, benchmark)
```

## grcwa · NumPy CPU

```python
from rcwa_cv.solvers.grcwa_cpu import probe

if benchmark["comparisons"]["grcwa"]:
    print(probe())
results["grcwa"] = solve_sweep("grcwa", spec, benchmark)
```

## Lumerical RCWA

The cell uses the licensed FDTD application through its Python API. It reuses one session and applies the numerical mode and settings recorded for each eligible point.

```python
from rcwa_cv.solvers.lumerical import probe

if any(benchmark["comparisons"].values()):
    print(probe())
results["lumerical"] = solve_sweep("lumerical", spec, benchmark)
```
